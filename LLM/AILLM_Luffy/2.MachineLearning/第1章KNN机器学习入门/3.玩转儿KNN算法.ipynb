# 4.1、分类

### 4.1.1、数据介绍

- 案例介绍，鸢尾花分类

- 气候不同，造就性状不同，统计了鸢尾花的关键特征数据：花萼长度、花萼宽度、花瓣长度、花瓣宽度。

- 植物学家划分为：
    - setosa（中文名：山鸢尾）
    - versicolor（中文名：杂色鸢尾）
    - virginica（中文名：弗吉尼亚鸢尾）

- 鸢尾花数据集中每个样本包含4个特征（花萼长度、花萼宽度、花瓣长度、花瓣宽度），用于对鸢尾花进行分类。

### 4.1.2、KNN算法建模

- 加载数据
- 数据预处理，拆分
- 声明算法，给定超参数
- 训练算法，拟合数据，归纳规律
- 算法应用，样本预测

#### 4.1.2.1、数据加载

```python
from sklearn import datasets # 数据集
from sklearn.neighbors import KNeighborsClassifier # 分类
# X 表示数据，y目标值
X,y = datasets.load_iris(return_X_y=True)
display(X,y)
```

```python
X.shape
```

#### 4.1.2.2、数据拆分

```python
# train训练数据，test测试数据
from sklearn. model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, # 拆分数据
test_size =0.2, # 拆分比例，测试数据20%30个样本
random_state=1024) # 固定了随机状态
display(X_train.shape, X_test.shape, y_train.shape, y_test.shape)
```

#### 4.1.2.3、算法建模

```python
model = KNeighborsClassifier(n_neighbors=5) #给定5个邻居
model.fit(X_train,y_train) #建模、拟合、训练
```

#### 4.1.2.4、算法预测

```python
y_ = model.predict(X_test)
y_ #预测值
```

```python
print('保留测试数据，真实值是：\n', y_test)
```

```python
# 准确率
(y_==y_test).mean()
```

```python
model.score(X_test,y_test)
```
