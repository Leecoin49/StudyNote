## 1.6 设定刻度

plt.xticks()和plt.yticks()方法

- 需指定刻度值和刻度名称 plt.xticks([刻度列表],[名称列表])

- 支持fontsize、rotation、color等参数设置

```python
plt.plot(x, y)

# 刻度是数字
# plt.xticks（选择的刻度值列表，刻度值对应的刻度标签列表）
_ = plt.xticks([0, np.pi/2, np.pi, 3*np.pi/2, 2*np.pi], ['0', 'π/2', 'π', '3π/2', '2π'])
_ = plt.yticks([-1, 0, 1], ['min', 0, 'max'])
plt.show()
```

```python
axes = plt.subplot(111)
#设置刻度的方法
_ = axes.set_xticks([0, np.pi/2, np.pi, 3*np.pi/2, 2*np.pi])
#设置刻度标签的方法
_ = axes.set_xticklabels(['0','π/2','π','3π/2','2π'])
_ = axes.set_yticks([-1, 0, 1])
_ = axes.set_yticklabels(['min','0','max'])
```

### 1.10.2 loc参数

- loc参数用于设置图例标签的位置，一般在legend函数内
- matplotlib已经预定义好几种数字表示的位置

| 字符串 (String) | 数值 (Value) |
| :--- | :--- |
| `best` | 0 |
| `upper right` | 1 |
| `upper left` | 2 |
| `lower left` | 3 |
| `lower right` | 4 |
| `right` | 5 |
| `center left` | 6 |
| `center right` | 7 |
| `lower center` | 8 |
| `upper center` | 9 |
| `center` | 10 |


```python
plt.figure(figsize=(10, 4))
ax1 = plt.subplot(121)
plt.plot(x, y1, color='red', label='sin')
plt.legend(loc='upper left')
ax2 = plt.subplot(1,2,2)
plt.plot(x, y2, color='green', label='cos')
plt.legend(loc='upper left')

plt.show()
```

loc参数可以是2元素的元组，表示图例左下角的坐标
- [0,0]	左下
- [0,1]	左上
- [1,0]	右下
- [1,1]	右上

```python
plt.plot(x, y1, label='sin')
plt.plot(x, y2, label='cos')
plt.legend(loc=[0.45, 1.1], ncol=2)
plt.show()
```

# 2 添加注释

添加文字的方法：

| pyplot函数 | API方法 | 描述 |
| :--- | :--- | :--- |
| `text()` | `mpl.axes.Axes.text()` | 在 Axes 对象的任意位置添加文字 |
| `figtext()` | `mpl.figure.Figure.text()` | 在 Figure 对象的任意位置添加文字 |
| `suptitle()` | `mpl.figure.Figure.suptitle()` | 为 Figure 对象添加中心化的标题 |
| `annotate()` | `mpl.axes.Axes.annotate()` | 为 Axes 对象添加注释（箭头可选） |


所有的方法会返回一个matplotlib.text.Text对象

## 2.1 图形内的文字

text()

```python
x = np.linspace(0, 2*np.pi, 20)
y = np.sin(x)
```

```python
figure = plt.figure(figsize=(10, 4), facecolor='lightgreen')
ax1 = plt.subplot(1,2,1)
plt.plot(x, y)
#x，y	表示文字在画布中的位置信息，使用画布刻度值来描述的
plt.text(x=3, y=0, s='dancer')
#ax1.text()

ax2 = plt.subplot(1,2,2)
plt.text(x=0.5, y=0.5, s='lucy', fontdict={'color':'green', 'fontsize':13})

# 添加画板标题
# x, y 描述文字绘制的位置信息，信息使用相对位置表示
figure.suptitle(t='sup title', x=0.5, y=1.0)

plt.show()
```

## 2.2 画板上添加文字

```python
figure = plt.figure(figsize=(10, 5))
ax1 = plt.subplot(3,1,1)
ax2 = plt.subplot(2,1,2)

x = np.linspace(-np.pi, np.pi, 50)
y = np.cos(x)

ax2.plot(x, y, color='red')

# ax2.annotate()
plt.annotate(xy=(0,1), text='max point', xytext=(0, 0), arrowprops={
    # 'arrowstyle':'simple',
    'color': 'red',
    'width': 3,
    'headlength': 5,
    'headwidth': 6,
    'shrink': 0.1
})

# x, y相对位置的描述
figure. text(x=0.4, y=0.55, s='this is a point', fontdict={
    'color':'green',
    'fontsize':18
})

plt.show()
```
