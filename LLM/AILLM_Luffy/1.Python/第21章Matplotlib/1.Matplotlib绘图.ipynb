# 2 添加注释

添加文字的方法：

| pyplot函数 | API方法 | 描述 |
| :--- | :--- | :--- |
| `text()` | `mpl.axes.Axes.text()` | 在 Axes 对象的任意位置添加文字 |
| `figtext()` | `mpl.figure.Figure.text()` | 在 Figure 对象的任意位置添加文字 |
| `suptitle()` | `mpl.figure.Figure.suptitle()` | 为 Figure 对象添加中心化的标题 |
| `annotate()` | `mpl.axes.Axes.annotate()` | 为 Axes 对象添加注释（箭头可选） |


所有的方法会返回一个matplotlib.text.Text对象

## 2.1 图形内的文字

text()

```python
x = np.linspace(0, 2*np.pi, 20)
y = np.sin(x)
```

```python
figure = plt.figure(figsize=(10, 4), facecolor='lightgreen')
ax1 = plt.subplot(1,2,1)
plt.plot(x, y)
#x，y	表示文字在画布中的位置信息，使用画布刻度值来描述的
plt.text(x=3, y=0, s='dancer')
#ax1.text()

ax2 = plt.subplot(1,2,2)
plt.text(x=0.5, y=0.5, s='lucy', fontdict={'color':'green', 'fontsize':13})

# 添加画板标题
# x, y 描述文字绘制的位置信息，信息使用相对位置表示
figure.suptitle(t='sup title', x=0.5, y=1.0)

plt.show()
```

## 2.2 画板上添加文字

```python
figure = plt.figure(figsize=(10, 5))
ax1 = plt.subplot(3,1,1)
ax2 = plt.subplot(2,1,2)

x = np.linspace(-np.pi, np.pi, 50)
y = np.cos(x)

ax2.plot(x, y, color='red')

# ax2.annotate()
plt.annotate(xy=(0,1), text='max point', xytext=(0, 0), arrowprops={
    # 'arrowstyle':'simple',
    'color': 'red',
    'width': 3,
    'headlength': 5,
    'headwidth': 6,
    'shrink': 0.1
})

# x, y相对位置的描述
figure. text(x=0.4, y=0.55, s='this is a point', fontdict={
    'color':'green',
    'fontsize':18
})

plt.show()
```

## 4.3 直方图

【直方图的参数只有一个x！！！不像条形图需要传入x,y】

hist()的参数

- bins

    可以是一个bin数量的整数值，也可以是表示bin的一个序列。默认值为10
- normed 现在叫density

    如果设为True，直方图的值将进行归一化处理，形成概率密度，默认值为False
- color

    指定直方图的颜色，可以是单一颜色值或颜色的序列。如果指定了多个数据集合，颜色序列将会设置为相同的顺序，如果未指定，将会使用一个默认的线条颜色
- orientation

    通过设置orientation为horizontal创建水平直方图。默认值为vertical

```python
score = np. random. randint(0,100, size=100)
score
```

```python
plt.hist(score)
plt.show()
```

```python
pd. Series(score). plot(kind='kde')
plt.show()
```

```python
# bins 整数 n 表达要把统计数据离散化，进行n等分的划分
#划分的区间是根据绘制的绘图来进行计算的 min max 分成n等份
plt.hist(score, bins=2, width=30, color='red', orientation='vertical', density=True)

# 密度图像 0-1
pd.Series(score).plot(kind='kde')
plt.show()
```

## 4.7饼图

【饼图也只有一个参数x！】

pie()

饼图适合展示各部分占总体的比例，条形图适合比较各部分的大小

- labels参数设置每一块的标签；
- labeldistance参数设置标签距离圆心的距离（比例值，只能设置一个浮点小数）
- autopct参数设置比例值的显示格式（%、%1.1%）；
- pctdistance参数设置比例值文字距离圆周的距离
- explode参数设置每一块顶点距圆形的长度（比例值、列表）；
- colors参数设置每一块的颜色（列表）；
- shadow参数为布尔值，设置是否绘制阴影
- startangle参数设置讲扇形起始角度
