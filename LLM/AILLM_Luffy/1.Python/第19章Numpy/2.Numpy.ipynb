## 1.5 ndarray的级联和切分

### 1.5.1 级联

级联的注意事项：
- 级联的参数是列表：一定要加中括号或小括号
- 维度必须相同
- 形状相符
- 级联的方向默认是shape这个tuple的第一个值所代表的维度方向
- 可通过axis参数改变级联的方向

```python
a1 = np. random. randint(0, 10, size=(3,4))
a2 = np. random. randint(10, 20, size=(3,4))
```

```python
display(a1, a2)
```

```python
np.concatenate((a1, a2), axis=0)
```

```python
a3 = np.random.randint(20,30,size=(3,))
a3
```

```python
# ValueError: all the input arrays must have same number of dimensions
# 参与级联的数组必须维度一致
np.concatenate(a1, a3)
```

```python
a4 = np.random.randint(30, 40, size=(3,3))
a4
```

```python
a1
```

```python
# ValueError: all the input array dimensions except for the concatenation axis must match exactly, but along dimension 1, the array at index 0 has size 4 and the array at index 1 has size 3
# 在级联的维度上，必须长度一致
np.concatenate((a1, a4), axis=0)
```

```python
np.concatenate((a1, a4), axis=1)
```

np.hstack与np.vstack

```python
# 竖着连接
np.vstack((a1, a4))
```

```python
# 横着连接
np.hstack((a1, a4))
```

### 1.5.2 切分

与级联类似，三个函数完成切分工作：
- np.split
- np.vsplit
- np.hsplit

```python
arr = np.random.randint(0, 100, size=(6,5))
arr
```

```python
# indices_or_secions
# int 会按照指定的轴向，将数组切分成N等分，要求切分的方向上的长度能被N整除
a1, a2 = np.split(arr, indices_or_sections=2, axis=0)
display(a1, a2)
```

```python
np.split(arr, indices_or_sections=4, axis=0)
```

```python
# indices_or_sections [m,n] 表示的是 按照0:m, m:n, n:的切片逻辑对数组进行拆分【左闭右开区间】
np.split(arr, indices_or_sections=[2,3], axis=1)
```

```python
arr
```

h horizontal 横向
v vertical 纵向

```python
np.vsplit(arr, indices_or_sections=2)
```

```python
np.hsplit(arr, indices_or_sections=[2,3])
```

### 1.6.2 广播 broadcast
ndarray广播机制的两条规则

如果
1. 两个数组的后缘维度(trailing dimension, 即从末尾开始算起的维度)的轴长度相符
2. 或其中的一方的长度为1
则认为它们是广播兼容的。广播会在缺失和（或）长度为1的维度上进行。

例1：a = np.ones((2, 3)) b = np.arange(3)求a + b

```python
a = np.ones((2,3))
b = np.arange(3)
display(a, b)
```

```python
a + b
```

例2: a = np.ones((4,3,2)), b = np.random.randint(0,10,size=(3,2)),求a+b

```python
a = np.ones((4,3,2))
b = np.random.randint(0, 10, size=(3,2))
display(a, b)
```

```python
a + b 
```

例3: a = np. arange(3). reshape((3, 1)), b = np. arange(3), 求a+b

```python
a = np.arange(3).reshape((3,1))
a
```

```python
b = np.arange(3)
b
```

```python
a + b
```

```python
# ndarray可以跟任意的整数进行广播运算
a + 1
```
