一个基于python的扩展库

提供了各种高级数据编程工具，如矩阵运算、向量运算、快速筛选、I/O操作，博里叶变换、线性代数，随机数等
## 1.2 ndarray属性
1. ndim:维度
2. shape:形状(各维度的长度)
3. size:总长度
4. dype:元素类型

### 1.3.2 强制类型统一

numpy设计初衷时用于运算的，所以对数据类型进行统一优化

注意:
- numpy默认ndarray的所有元素的类型是相同的

- 如果传进来的列表中包含不同的类型，则统一为同一类型，优先级:str>float>int

```python
list1 = [1, 2, 3, 4.12]
list1
```

```python
arr3 = np.array(list1)
```

```python
arr3.dtype
```

```python
list2 = [1, 2, 3, 3.14, 'object']
arr4 = np.array(list2)
```

```python
# U 表示Unicode的简写形式
arr4.dtype
```

```python
type(arr2) # 查看arr2对象本身的类型
```

```python
arr2.dtype # 查看arr2内部的元素类型，由于ndarray的元素类型是强制统一的，所以只返回一个类型
```
