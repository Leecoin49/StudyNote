## 1.4 异常值处理

体育比赛中，最后算分的时候总会去掉一个最高分，去掉一个最低分，将剩下的分数进行平均。因为从一个非常高的异常值可能会造成对最后结果产生比较大的影响。所以，异常值处理，在统计数据的过程中是必然要介入的一个操作

### 1.4.2 异常值的认定

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline
```

不符合业务逻辑的异常值

使用均值和标准差进行判断

数据要先满足服从标注正态分布

abs(X) > 3 * X.std()

```python
df = pd.DataFrame(data=np.random.randn(10000, 4), columns=list('ABCD'))
```

```python
df
```

```python
sns.histplot(df.A)
```

```python
df.std()
```

```python
df
```

```python
exp_condition = (np.abs(df) > df.std() * 3).any(axis=1)
```

```python
exp_index = df.loc[exp_condition].index
```

```python
# 过滤 写出正常值的过滤条件
normal_condition = (np.abs(df) < df.std() * 3).all(axis=1)
df.loc[normal_condition]
```

```python
# 删除指定索引的值
# index <==> labels axis
# df.drop(index=exp_index)
# columns 指定列索引 <==> labels axis=1
df.drop(labels=exp_index, axis=0)
```

使用上四中位数和下四中位数进行异常值判定

```python
order = pd.read_excel('电商用户数据.xlsx')
```

```python
order.info()
```

```python
order.describe(percentiles=[.99])
```

```python
sns.boxplot(x=order['实付金额'])
```

```python
# 25 75

Q1, Q3 = np.quantile(order['实付金额'], q=[0.25, 0.75])

IQR = Q3 - Q1
```

```python
BOTTOM = Q1 - 1.5 * IQR
UPPER = Q3 + 1.5 * IQR
```

```python
(order['实付金额'] > UPPER).values
```

```python
condition = (order['实付金额'] > UPPER).values | (order['实付金额'] < BOTTOM).values
```

```python
exp_index = order.loc[condition].index
```

```python
order.drop(index=exp_index).describe()
```
