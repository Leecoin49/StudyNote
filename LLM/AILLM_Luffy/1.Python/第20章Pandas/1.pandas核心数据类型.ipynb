# 1. Pandas 核心数据类型

## 1.1 什么是Pandas

- Pandas是Python语言的一个扩展程序库，用于数据分析。
- Pandas是一个开放源码、BSD许可的库，提供高性能、易于使用的数据结构和数据分析工具。
- Pandas名字衍生自术语"panel data"（面板数据）和"Python data analysis"（Python数据分析）。
- Pandas一个强大的分析结构化数据的工具集，基础是Numpy（提供高性能的矩阵运算）。
- Pandas可以从各种文件格式比如CSV、JSON、SQL、Microsoft Excel导入数据。
- Pandas可以对各种数据进行运算操作，比如归并、再成形、选择，还有数据清洗和数据加工特征。

Pandas的主要数据结构是Series（一维数据）与DataFrame（二维数据），这两种数据结构足以处理金融、统计、社会科学、工程等领域里的大多数典型用例

## 1.3 DataFrame

DataFrame是一个二维的表格型数据结构，可以看做是由Series组成的字典(共用同一个索引)。

DataFrame由按一定顺序排列的【多列】数据组成，每一列的数类型可能不同。

设计初衷是将Series的使用场最从一维拓展到多维，DataFrame既有行索引，也有列索引。

- 行索引:index
- 列索引:columns
- 值:values(numpy的二维数组)

```python
list1 = [1, 2, 'name']

s1 = pd.Series(data=list1)
```

### 1.4.5 where() and mask()

where:满足con条件的保留，其他用other填充

mask:不满足con条件的保留，其他用other填充

```python
df = pd.DataFrame(data=np.random.randint(0,100,size=(5,4)), columns=list('ABCD'))
df
```

```python
df > 60
```

```python
condition = df > 60
```

```python
condition = df['A'] > 60
```

```python
df.where(cond=df > 60, other='不及格')
```

```python
df.where(cond=df['A'] > 60, other='其他')
```

```python
df.where(cond=df>60,other=pd.Series(data=[100,200],index=['A','B']),axis=1)
```

```python
df.where(cond=df>60,other=pd.Series(data=[100,200],index=[1,2]),axis=0)
```

```python
df.mask(cond=df > 60,other='及格')
```

```python
df.mask(cond=df > 60, other=pd.Series(data=[100, 200, 300, 400], index=['A','B', 'C', 'D']), axis=1)
```
