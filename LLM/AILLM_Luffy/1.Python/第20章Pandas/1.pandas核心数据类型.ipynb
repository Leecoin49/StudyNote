# 1. Pandas 核心数据类型

## 1.1 什么是Pandas

- Pandas是Python语言的一个扩展程序库，用于数据分析。
- Pandas是一个开放源码、BSD许可的库，提供高性能、易于使用的数据结构和数据分析工具。
- Pandas名字衍生自术语"panel data"（面板数据）和"Python data analysis"（Python数据分析）。
- Pandas一个强大的分析结构化数据的工具集，基础是Numpy（提供高性能的矩阵运算）。
- Pandas可以从各种文件格式比如CSV、JSON、SQL、Microsoft Excel导入数据。
- Pandas可以对各种数据进行运算操作，比如归并、再成形、选择，还有数据清洗和数据加工特征。

Pandas的主要数据结构是Series（一维数据）与DataFrame（二维数据），这两种数据结构足以处理金融、统计、社会科学、工程等领域里的大多数典型用例

## 1.2 Series

Series是一种类似于字典(dict-like)的一维数组(array-like)的对象，由下面两个部分组成:

- values:-组数据(ndarray类型)

- index:相关的数据索引标签

```python
import numpy as np

import pandas as pd

from pandas import Series, DataFrame
```

### 1.2.1 构造Series

1. From ndarray 使用ndarray构造 是一个引用对象

```python
arr1 = np.random.randint(0, 10, size=5)
arr1
```

```python
arr1[0] = 100
```

```python
s1 = pd.Series(data=arr1)
```

```python
# 注意：此处是修改了arr1[0]=100之后的输出
s1
```

```python
# index用来执行Series对象的显示索引，如果没有显示设置索引，则默认使用隐示（数组索引）索引填充
s2 = pd.Series(data=arr1, index=['a', 'b', 'c', 'd', 'e'])
s2
```

## 1.3 DataFrame

DataFrame是一个二维的表格型数据结构，可以看做是由Series组成的字典(共用同一个索引)。

DataFrame由按一定顺序排列的【多列】数据组成，每一列的数类型可能不同。

设计初衷是将Series的使用场最从一维拓展到多维，DataFrame既有行索引，也有列索引。

- 行索引:index
- 列索引:columns
- 值:values(numpy的二维数组)

```python
list1 = [1, 2, 'name']

s1 = pd.Series(data=list1)
```

```python
# Series和ndarray一样，会强制类型统一
s1
```

```python
s1.dtype
```


### 1.3.1 构造DataFrame

```python
# data-None 用于构造DataFrame的数据
# index=None, columms=None, 行列索引
# dtype=None，数据类型
# copy=False

data = np.random.randint(0, 100, size=(3, 4))
index = ['tom', 'lucy', 'alex']
columns = ['语文', '数学', '英语', '物理']
pd.DataFrame(data=data, index=index, columns=columns)
```

1. From dict of Series or dicts 使用一个由Series构造的字典或一个字典构造

2. From dict ofndarrays/lists 使用一个由列表或ndarray构造的字典构造

ndarrays必须长度保持一致

3. From alist of dicts 使用一个由字典构成的列表构造
