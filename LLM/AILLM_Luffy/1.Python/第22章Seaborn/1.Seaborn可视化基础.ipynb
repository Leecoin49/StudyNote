## 1.1 一、图表控制

了解seaborn如何对matplotlib输出的外观进行控制

导包

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import seaborn as sns
```

定义一个正弦函数，比较seaborn绘图和matplotlib绘图背景的区别

```python
x = np. linspace(0,2*np. pi,100)
y = np.sin(x)
plt.plot(x,y)
plt.show()
```

```python
def show_lines(x):
    for i in range(7):
        y = np.sin(x) + i
        plt.plot(x, y)
        plt.show()
```

```python
# 默认的Matplotlib样式
show_lines(x)
```

```python
plt.rcParams['axes.facecolor']='white'
```

```python
show_lines(x)
```

使用sns.set()来设置seaborn默认的画布风格

```python
# 全局配置
sns.set()
```

```python
show_lines(x)
plt.title('sin x')
```

1.1.1 样式控制：
1. axes_style：获取设置背景样式的属性字典
2. set style(style, rc)：设置背景样式

style可选的5个seaborn默认主题
- darkgrid黑色网格（默认）
- whitegrid白色网格
- dark黑色背景
- white白色背景
- ticks有刻度线的白背景
- rc背景样式的属性字典设置接口

sns.axes_style()函数可以返回可以设置的属性字典

```python
sns.axes_style()
```

```python
# 全局配置 画布风格
##全局配置画布风格
sns.set_style(style='darkgrid', rc={'axes.facecolor':'#EAEAF2'})
```

```python
show_lines(x)
```

```python
# 临时配置
with sns.axes_style(style='ticks'):
    show_lines(x)
```

### 1.1.2 边框控制 despine()
- 默认删除上方和右方的边框
- offset两坐标轴分开的距离
- trim True False限制预留的边框范围，没有图像的地方会舍弃
- top right bottom left (True/False)指定是否删除对应的边框
- 注意边框控制的代码应放在绘图之后

```python
sns.set_style(style='darkgrid', rc={
    'axes.edgecolor': 'black',
    'axes.spines.left': False
})
show_lines(x)
```

```python
# plt.figure(figsize=(8,8))
show_lines(x)
plt.axis('equal')

# dispine要在绘图函数后生效，设置为True，即为隐藏对应的边框
sns.despine(left=False, top=False, right=False, bottom=False, offset=10, trim=5)
```
