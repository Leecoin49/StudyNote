## 1 Transformer的诞生
2018年10月，Google发出一篇论文《BERT: Pre-training of Deep Bidirectional Transformers for Language Understanding》，BERT模型横空出世，并横扫NLP领域11项任务的最佳成绩！

论文地址：https://arxiv.org/pdf/1810.04805.pdf

而在BERT中发挥重要作用的结构就是Transformer，之后又相继出现XLNET、roBERT等模型也取了BERT，但是他们的核心没有变，仍然是：Transformer。

## 2 Transformer的优势
相比之前占领市场的LSTM和GRU模型，Transformer有两个显著的优势：

1. Transformer能够利用分布式GPU进行并行训练，提升模型训练效率。
2. 在分析预测更长的文本时，捕捉更长的语义关联效果更好。

## 💡 第一性原理：为什么 Transformer 能取代 RNN/LSTM？

### RNN 的根本瓶颈

回顾 RNN 的处理方式：读完词1才能读词2，读完词2才能读词3... 像一条单行道。

```
RNN: 词₁ → 词₂ → 词₃ → ... → 词ₙ
     └── 必须等前面算完才能往后走 ──┘
```

这导致两个致命问题：
1. **无法并行**：GPU有几千个核心，但RNN只能一次用一个
2. **长距离衰减**：梯度消失，前面的词传递100步后信息几乎为零

### Transformer 的第一性洞察

> **如果RNN的"循环"是问题的根源，那就彻底丢掉循环！**

Transformer 的核心思想：
```
RNN:  词₁→词₂→词₃    ← 串行，一步接一步
Transformer:  每个词同时看到所有其他词    ← 并行，一步到位
```

这就是**自注意力机制（Self-Attention）**替代循环的动机——让句子中任意两个词之间直接建立联系，距离恒为 O(1)。

### 费曼类比：RNN vs Transformer

```
RNN       = 传话游戏，100个人排成一排，只能和前后的人说话
Transformer = 圆桌会议，所有人同时坐在一张桌子上，每个人都能直接和任何其他人对话
```

- 传话游戏传到后面信息严重失真（梯度消失）
- 圆桌会议每个人直接交流，不会丢信息

> **这就是 Transformer 的两大核心优势：并行训练快 + 长距离语义捕捉好。**

# 3 Transformer的市场

在著名的SOTA机器翻译榜单上，几乎所有排名靠前的模型都使用Transformer。

其基本上可以看作是工业界的风向标，市场空间自然不必多说！
